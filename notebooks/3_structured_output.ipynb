```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
text="hello my  name is sneha pandey and my email is sneha.pandey@.com and my phone number is 1234567890"
```

```python
res = llm.invoke(
    f"Please extract the name, email and phone number from the text: {text}"
)
```

```python
res.content
```

```python
from pydantic import BaseModel, Field
from typing import List

class ResponseStruct(BaseModel):
    name: str = Field(..., description="The name extracted from the text")
    email: str = Field(..., description="The email extracted from the text")
    phone_number: str = Field(..., description="The phone number extracted from the text")


structured_llm = llm.with_structured_output(ResponseStruct)

res = structured_llm.invoke(
    f"Please extract the name, email and phone number from the text: {text}"
)

print(res)
```

```python
from pydantic import BaseModel, Field

class Movies(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    director: str = Field(..., description="The director of the movie")
    genre: str = Field(..., description="The genre of the movie")


class AllowMovies(BaseModel):
    movies: list[Movies]
```

```python
structured_llm = llm.with_structured_output(
    AllowMovies,
    method="json_schema"
)

res = structured_llm.invoke(
    "Give me information about 3 movies. Include title, year, director and genre."
)

print(res)
```
