```python
#dynamic prompt generation
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGroq(model="openai/gpt-oss-120b")
out=StrOutputParser()
```

```python
#my  self chain node
def transform_case(result: str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Answer in {language}."),
    ("human", "{query}")
])
```

```python
final_prompts = prompts.format_messages(
    language="tamil",
    query="hii how are you"
)
final_prompts
```

```python
res=llm.invoke(final_prompts)
print(res.content)
```

```python
res=llm.invoke(prompts.invoke({"language": "tamil", "query": "hii how are you"}))
res.content
```

```python
chains=prompts | llm | out | transform_case
res=chains.invoke({"language": "english", "query": "hii how are you"})
print(res)
```
