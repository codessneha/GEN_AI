```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """it will add two numbers and return the result
    Args:
    a:number 1
    b:number 2
    """
    return a + b
```

```python
@tool
def multiply_numbers(a: int, b: int):
    """it will multiply two numbers and return the result
    Args:
    a:number 1
    b:number 2
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 10})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a helpful assistant that can perform addition and multiplication by using tools."
    
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"Please add 5 and 10 and multiply the result by 2."}]

})
```

```python
for res in response["messages"]:
    print(res)
    print()
```
