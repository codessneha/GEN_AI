```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
res=llm.invoke("hii im sneha")
print(res.content)
```

```python
res=llm.invoke("what is my name")
print(res.content)
```

```python
prompts = [
    {"role": "user", "content": "hi, my name is Sneha"},
    {"role": "assistant", "content": "Hi Sneha! How can I assist you today?"},
    {"role": "user", "content": "what is my name?"},
]

res=llm.invoke(prompts)
print(res.content)
```

```python
#manage histor
history = []
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit"]:
        break
    history.append({"role": "user", "content": query})
    print("User:", query)
    res = llm.invoke(history)
    history.append({"role": "assistant", "content": res.content})
    print("Assistant:", res.content)
```
